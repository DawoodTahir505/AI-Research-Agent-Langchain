```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_classic.agents import create_react_agent, AgentExecutor
from langchain_classic import hub
import requests
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from groq import Groq

client = Groq(api_key=os.getenv("GROQ_API_KEY"))
for m in client.models.list().data:
    print(m.id)
```

```python
search_tool_duck = DuckDuckGoSearchRun()
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b', temperature=0)
```

```python
search_tool_duck.invoke('what is data science?')
```

```python
@tool
def search_tool(input:str) ->str:
    """this tool will help you to search and research"""
    return search_tool_duck.invoke(input)
```

```python
from langsmith import Client

client = Client()
prompt = client.pull_prompt("hwchase17/react", dangerously_pull_public_prompt=True)
```

```python
agent = create_react_agent(
    llm=llm,
    tools=[search_tool],
    prompt=prompt
)
```

```python
agent_executor = AgentExecutor(agent=agent, tools=[search_tool], verbose=True)
```

```python
agent_executor.invoke({"input":"research and answer in 3 word what is data science?"})
```
